```python
import pandas as pd
import matplotlib.pyplot as plt
import torch
import numpy as np

data = pd.read_csv("../data/synthetic_data.csv")

# Separate the two classes
class_0 = data[data["output"] == 0]
class_1 = data[data["output"] == 1]

plt.scatter(
    class_0["X1"],
    class_0["X2"],
    label="Class 0"
)

plt.scatter(
    class_1["X1"],
    class_1["X2"],
    label="Class 1"
)

plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Synthetic Circular Classification Dataset")
plt.legend()
plt.axis("equal")
plt.show()

# ---------------------------------------------------------
# True decision boundary
# x1^2 + x2^2 = r^2
# ---------------------------------------------------------

radius = 1.1968   # Change this if your generator used another radius

theta = np.linspace(0, 2 * np.pi, 500)

circle_x = radius * np.cos(theta)
circle_y = radius * np.sin(theta)

# ---------------------------------------------------------
# Plot
# ---------------------------------------------------------

plt.figure(figsize=(8, 6))

# Class 0
plt.scatter(
    class_0["X1"],
    class_0["X2"],
    s=30,
    alpha=0.7,
    label="Class 0"
)

# Class 1
plt.scatter(
    class_1["X1"],
    class_1["X2"],
    s=30,
    alpha=0.7,
    label="Class 1"
)

# True boundary
plt.plot(
    circle_x,
    circle_y,
    linestyle="--",
    linewidth=2,
    label="True boundary"
)

plt.xlabel("$X_1$")
plt.ylabel("$X_2$")

plt.title("Nonlinear Dataset\n(Circle)")

plt.legend()

# Make the circle actually look circular
plt.axis("equal")

plt.grid(alpha=0.3)

plt.show()
```
